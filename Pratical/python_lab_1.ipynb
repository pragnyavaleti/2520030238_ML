Program 4: Lists and Indexing
Write a Python program to create a list and access its elements using indexing.
Write a program to display the first, second, and last elements of a list.
Write a program to demonstrate list slicing.
Write a program to add a new element to an existing list using the append() method.
Write a program to display the total number of elements in a list.

```python
fruits = ["Apple", "Banana", "Mango", "Orange", "Grapes"]

# Access elements using indexing
print("List =", fruits)
print("First element =", fruits[0])
print("Second element =", fruits[1])
print("Last element =", fruits[-1])

# List slicing
print("First three elements =", fruits[:3])
print("Last two elements =", fruits[3:])

# Add a new element
fruits.append("Pineapple")
print("After append =", fruits)

# Total number of elements
print("Total elements =", len(fruits))
```

Program 5: Dictionaries
Write a Python program to create a dictionary and display all its key-value pairs.
Write a program to access the value associated with a specific key in a dictionary.
Write a program to iterate through a dictionary using the items() method.
Write a program to store different machine learning models in a dictionary and display their names and objects.

```python
# Create a dictionary
student = {
    "Name": "Pragnya",
    "Roll No": 101,
    "CGPA": 9.2
}

# Display all key-value pairs
print("Student Dictionary =", student)

# Access value using a key
print("Student Name =", student["Name"])

# Iterate through the dictionary
print("\nDictionary Elements")
for key, value in student.items():
    print(key, ":", value)

# Machine Learning Models Dictionary
models = {
    "Linear Regression": "Model 1",
    "Decision Tree": "Model 2",
    "Random Forest": "Model 3",
    "SVM": "Model 4"
}

print("\nMachine Learning Models")
for name, obj in models.items():
    print(name, "=", obj)
```

Program 7: Loops
Write a Python program to display all elements in a list using a for loop.
Write a program to calculate the average of numbers stored in a list.
Write a program to display only the values greater than a given threshold using a loop.
Write a program to find the maximum and minimum values in a list using loops.
Write a program to count the number of elements satisfying a given condition using a loop.


```python
# Program 7: Loops

numbers = [10, 20, 30, 40, 50]

# Display all elements
print("List elements:")
for i in numbers:
    print(i)

# Average of numbers
sum = 0
for i in numbers:
    sum = sum + i
print("Average =", sum / len(numbers))

# Values greater than threshold
threshold = 25
print("Values greater than", threshold)
for i in numbers:
    if i > threshold:
        print(i)

# Maximum and Minimum
max = numbers[0]
min = numbers[0]

for i in numbers:
    if i > max:
        max = i
    if i < min:
        min = i

print("Maximum =", max)
print("Minimum =", min)

# Count elements greater than 25
count = 0
for i in numbers:
    if i > 25:
        count = count + 1

print("Count =", count)
```
